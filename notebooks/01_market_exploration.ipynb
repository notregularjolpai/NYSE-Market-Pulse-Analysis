# NYSE Market Pulse — Exploratory Notebook

**Live price analysis only when the downloaded price cache exists.** The fallback uses *synthetic*, fabricated demo prices.

Run from repository root: `jupyter notebook notebooks/01_market_exploration.ipynb`

```python
from pathlib import Path
import sys
ROOT = Path.cwd() if (Path.cwd() / "marketpulse").exists() else Path.cwd().parent
sys.path.insert(0, str(ROOT))
from marketpulse.data import load_prices, make_demo_prices
from marketpulse.analytics import performance_table, wealth_index, market_breadth, drawdowns
from marketpulse.universe import STOCKS, BENCHMARK
```

```python
cache = ROOT / "data" / "processed" / "adjusted_close.csv"
if cache.exists():
    prices = load_prices(cache)
    print("HISTORICAL DATA: Yahoo Finance source; latest observation:", prices.index[-1].date())
else:
    prices = make_demo_prices()
    print("SYNTHETIC DEMO — THESE ARE NOT REAL STOCK-MARKET PRICES")
prices.tail()
```

```python
metrics = performance_table(prices)
metrics[["Sector", "Annualized Return (CAGR)", "Annualized Volatility", "Sharpe Ratio", "Max Drawdown", "Beta vs SPY"]].round(3)
```

```python
axes = wealth_index(prices)[["JPM", "XOM", "KO", BENCHMARK, "EQUAL_WEIGHT"]].plot(figsize=(11, 5), title="Growth of $100 — " + ("HISTORICAL" if cache.exists() else "SYNTHETIC DEMO"))
axes.set_ylabel("Wealth ($)")
```

```python
dd = drawdowns(wealth_index(prices))
(dd[[BENCHMARK, "EQUAL_WEIGHT"]] * 100).plot(figsize=(11, 4), title="Drawdown comparison (%)");
```

```python
breadth = market_breadth(prices)
breadth.plot(figsize=(11, 4), title="Share of 11 selected NYSE stocks with positive 63-day return (%)");
```
